## Tujuan
Mengimplementasikan A* pada graph kecil dan membandingkan beberapa heuristic.

```python
import heapq

graph = {
    'A': {'B': 2, 'C': 5},
    'B': {'C': 1, 'D': 4},
    'C': {'D': 1, 'E': 7},
    'D': {'E': 2},
    'E': {}
}
heuristic = {'A': 5, 'B': 4, 'C': 3, 'D': 2, 'E': 0}

def astar(graph, start, goal, h):
    frontier = [(h[start], 0, start, [start])]
    best_g = {start: 0}
    expanded = []
    while frontier:
        f, g, node, path = heapq.heappop(frontier)
        if g != best_g.get(node):
            continue
        expanded.append(node)
        if node == goal:
            return path, g, expanded
        for nxt, cost in graph[node].items():
            new_g = g + cost
            if new_g < best_g.get(nxt, float('inf')):
                best_g[nxt] = new_g
                heapq.heappush(frontier, (new_g + h[nxt], new_g, nxt, path + [nxt]))
    return None, float('inf'), expanded

path, cost, expanded = astar(graph, 'A', 'E', heuristic)
path, cost, expanded
```

### Eksperimen 1
Catat path, total cost, dan urutan node yang dieksplor. Jelaskan peran **g(n)**, **h(n)**, dan **f(n)=g(n)+h(n)**.

```python
h_zero = {k: 0 for k in heuristic}
for name, h in [('heuristic_awal', heuristic), ('h=0', h_zero)]:
    path, cost, expanded = astar(graph, 'A', 'E', h)
    print(name, '=>', path, 'cost=', cost, 'expanded=', expanded)
```

##### Hasilnya : 
    - Path : A-> B-> C-> D-> E
    - Cost : 6 (2+1+1+2)
    - Urutan yang di ekspor = A,B,C,D,E

#### Peran g(n),h(n) dan f(n)=g(n)+h(n)

    -g(n) adalah biaya nyata dari titik awal sampai node saat ini 

    -h(n) perkiraan biaya dari node saat ini menuju tujuan(heuristic)

    -f(n)=g(n)+h(n) adalah perkiraan total biaya jalur yang lewat node n.

### Eksperimen 3

```python
h_alternatif = heuristic.copy()
h_alternatif['A'] = 9 #nilai A diubah dari 5 menjadi 9

path_alt, cost_alt, expanded_alt = astar(graph, 'A', 'E', h_alternatif)
print('heuristic_alternatif =', h_alternatif)
print('path =', path_alt)
print('cost =', cost_alt)
print('expanded =', expanded_alt)
```

## Refleksi akhir

1. Apa hasil utama yang Anda amati?
Hasil eksperimen menunjukkan bahwa A* menemukan rute A → B → C → D → E dengan total biaya 6. Ketika menggunakan heuristic awal maupun h = 0, rute optimal, total biaya, dan urutan node yang dieksplorasi tetap sama, yaitu A, B, C, D, dan E.

2. Mengapa hasil tersebut terjadi menurut konsep di Main Deck?
A* menggunakan fungsi evaluasi f(n) = g(n) + h(n), dengan g(n) sebagai biaya aktual dari titik awal dan h(n) sebagai perkiraan biaya menuju tujuan. Heuristic membantu algoritma memperkirakan arah pencarian. Ketika h(n) = 0, A* hanya mempertimbangkan biaya aktual dan berperilaku seperti Dijkstra. Pada graph kecil ini, kedua konfigurasi menghasilkan urutan eksplorasi yang sama.

3. Apa limitation eksperimen ini?
Eksperimen menggunakan graph kecil dengan lima node sehingga perbedaan kinerja heuristic belum terlihat jelas. Selain itu, hanya satu konfigurasi heuristic awal dan satu konfigurasi h = 0 yang diuji. Eksperimen ini belum membandingkan jumlah node yang dieksplorasi pada graph yang lebih besar atau menguji heuristic yang tidak admissible.

4. Eksperimen apa yang layak dilakukan berikutnya?
Eksperimen berikutnya dapat menggunakan graph yang lebih besar dengan lebih banyak jalur alternatif dan membandingkan heuristic admissible, heuristic yang terlalu tinggi, serta h = 0. Pengamatan dapat mencakup rute yang diperoleh, total biaya, jumlah node yang dieksplorasi, dan waktu eksekusi. Hasilnya dapat digunakan untuk memahami pengaruh kualitas heuristic terhadap efisiensi pencarian.
